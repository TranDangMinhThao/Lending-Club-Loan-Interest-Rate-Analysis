## 1. TIỀN XỬ LÝ DỮ LIỆU

Dữ liệu thực tế từ Lending Club có **151 cột** và hơn 2 triệu dòng, bao gồm nhiều vấn đề cần giải quyết trước khi mô hình hóa:

| Vấn đề | Rủi ro nếu bỏ qua |
|--------|-------------------|
| Nhiều cột không liên quan | Gây nhiễu, tăng thời gian tính toán (curse of dimensionality) |
| Giá trị thiếu (missing values) | Một số thuật toán không xử lý được NaN, gây lỗi hoặc bias |
| Ngoại lai (outliers) | Kéo lệch phân phối, ảnh hưởng nghiêm trọng đến hồi quy OLS |
| Biến phân loại (categorical) | Mô hình toán học chỉ nhận đầu vào dạng số |

Quy trình tiền xử lý của nhóm lần lượt là: **Load → Select → Clean → Encode**.


````python
# Load dữ liệu – dùng nrows để kiểm soát bộ nhớ, low_memory=False để đọc đúng kiểu
df_raw = pd.read_csv('data.csv', low_memory=False)

print(f' Kích thước dữ liệu gốc: {df_raw.shape[0]:,} dòng × {df_raw.shape[1]} cột')
print(f' Biến mục tiêu (int_rate):')
print(f'   Min = {df_raw["int_rate"].min():.2f}%  |  Max = {df_raw["int_rate"].max():.2f}%')
print(f'   Mean = {df_raw["int_rate"].mean():.2f}%  |  Std = {df_raw["int_rate"].std():.2f}%')
df_raw.head(3)
````

###  Insight – Khám phá ban đầu

- Bộ dữ liệu có **151 cột**, phần lớn là các chỉ số tài chính nội bộ (số dư tài khoản, lịch sử thanh toán…) không phù hợp để dự đoán vì chúng chỉ có sau khi khoản vay đã được giải ngân.
- `int_rate` dao động từ **5.31%** đến **30.99%**, trung bình **13.09%** (Std ≈ 4.83%). Phân phối lệch phải nhẹ cho thấy đa số khoản vay có lãi suất vừa phải, một số ít khoản vay rủi ro cao có lãi suất rất cao.
- Nhiều cột có **>50% giá trị thiếu** (các trường joint application, hardship, settlement…) → cần loại bỏ sớm để giảm nhiễu.
- **Chiến lược:** Chọn một tập đặc trưng có ý nghĩa tài chính thực sự, đã được công bố tại thời điểm tạo khoản vay (no data leakage).

````python
# ─── Chọn các đặc trưng có ý nghĩa tài chính & không bị data leakage ───────
# Chỉ giữ lại những biến đã biết TẠI THỜI ĐIỂM xét duyệt khoản vay
FEATURES = [
    # Thông tin khoản vay
    'loan_amnt',        # Số tiền vay
    'term',             # Kỳ hạn (36 hay 60 tháng)
    'installment',      # Số tiền trả mỗi tháng
    'grade',            # Hạng tín dụng nội bộ (A–G)
    'purpose',          # Mục đích vay

    # Thông tin người vay
    'emp_length',       # Thâm niên công việc
    'home_ownership',   # Tình trạng nhà ở
    'annual_inc',       # Thu nhập hàng năm
    'verification_status',  # Trạng thái xác minh thu nhập

    # Hồ sơ tín dụng
    'dti',              # Debt-to-Income ratio
    'fico_range_low',   # Điểm FICO (thấp)
    'fico_range_high',  # Điểm FICO (cao)
    'open_acc',         # Số tài khoản tín dụng đang mở
    'pub_rec',          # Số lần có hồ sơ công khai xấu
    'revol_bal',        # Số dư xoay vòng
    'revol_util',       # Tỷ lệ sử dụng hạn mức tín dụng xoay vòng
    'total_acc',        # Tổng số tài khoản tín dụng
    'delinq_2yrs',      # Số lần thanh toán trễ (2 năm qua)
    'inq_last_6mths',   # Số lần tra cứu tín dụng (6 tháng qua)
]

TARGET = 'int_rate'

df = df_raw[FEATURES + [TARGET]].copy()

# Tính fico_score trung bình từ fico_range_low và fico_range_high
df['fico_score'] = (df['fico_range_low'] + df['fico_range_high']) / 2

print(f'  Đã chọn {len(FEATURES)} đặc trưng + 1 biến mục tiêu')
print(f'   Kích thước sau khi chọn: {df.shape}')
print()

# Tóm tắt kiểu dữ liệu
dtype_summary = df.dtypes.value_counts()
print('Phân loại biến:')
print(f'   Số (numeric) : {dtype_summary.get("float64", 0) + dtype_summary.get("int64", 0)} biến')
print(f'   Chuỗi (object): {dtype_summary.get("object", 0)} biến — cần encoding')
````

### Lý do để chọn các đặc trưng này

**Nguyên tắc tránh Data Leakage:** Không được sử dụng các biến chỉ có *sau khi* khoản vay được phê duyệt (ví dụ: `total_pymnt`, `recoveries`, `last_pymnt_amnt`). Dùng những biến đó sẽ làm mô hình có R² rất cao giả tạo, nhưng hoàn toàn vô dụng trong thực tế.

**Lý do chọn từng nhóm:**

| Nhóm | Ý nghĩa tài chính |
|------|-------------------|
| `grade`, `fico_range_low/high` | Đây là yếu tố quyết định chính — Lending Club dùng grade để định giá lãi suất. Tương quan với `int_rate` đạt ~0.95 (grade) và ~-0.42 (fico). |
| `dti`, `revol_util` | Đo áp lực nợ của người vay — dti càng cao, `int_rate` càng tăng (corr ≈ 0.19, 0.20) |
| `inq_last_6mths` | Nhiều lần tra cứu tín dụng gần đây → dấu hiệu cần tiền gấp → rủi ro cao hơn (corr ≈ 0.24) |
| `annual_inc`, `home_ownership` | Phản ánh năng lực hoàn nợ của người vay |
| `term`, `loan_amnt` | Khoản vay lớn hơn / dài hạn hơn → rủi ro và lãi suất cao hơn |

**Đã loại bỏ:** 132 cột còn lại vì hoặc bị data leakage, hoặc thiếu dữ liệu quá nhiều (>50%), hoặc không có ý nghĩa kinh tế rõ ràng.


````python
# ─── Phân tích & xử lý giá trị thiếu ───────────────────────────────────────
print('=' * 50)
print(' PHÂN TÍCH GIÁ TRỊ THIẾU')
print('=' * 50)

missing = df.isnull().sum()
missing_pct = (missing / len(df) * 100).round(2)
missing_df = pd.DataFrame({'Số lượng': missing, 'Tỷ lệ (%)': missing_pct})
missing_df = missing_df[missing_df['Số lượng'] > 0].sort_values('Tỷ lệ (%)', ascending=False)
print(missing_df)
print()

# ─── Xử lý từng biến ────────────────────────────────────────────────────────

# 1. emp_length: 6.5% thiếu → điền bằng mode (giá trị phổ biến nhất)
emp_mode = df['emp_length'].mode()[0]
df['emp_length'].fillna(emp_mode, inplace=True)
print(f' emp_length: điền mode = "{emp_mode}"')

# 2. revol_util: <0.1% thiếu → điền bằng median (ít bị ảnh hưởng bởi outlier)
revol_median = df['revol_util'].median()
df['revol_util'].fillna(revol_median, inplace=True)
print(f' revol_util: điền median = {revol_median:.2f}%')

# Xác nhận
remaining_missing = df.isnull().sum().sum()
print(f'\n Số giá trị thiếu còn lại: {remaining_missing}')

# Trực quan hóa
fig, ax = plt.subplots(figsize=(7, 3))
ax.barh(['revol_util', 'emp_length'],
        [0.08, 6.5],   # tỷ lệ % trước xử lý
        color=['#4C9BE8', '#E87B4C'])
ax.set_xlabel('Tỷ lệ thiếu (%)')
ax.set_title('Missing Values trước khi xử lý', fontweight='bold')
for i, v in enumerate([0.08, 6.5]):
    ax.text(v + 0.05, i, f'{v}%', va='center', fontsize=10)
plt.tight_layout()
plt.show()
````

###  Chiến lược xử lý giá trị thiếu

Sau khi chọn lọc 19 đặc trưng, chỉ còn 2 biến có giá trị thiếu:

| Biến | Tỷ lệ thiếu | Chiến lược | Lý do |
|------|------------|------------|-------|
| `emp_length` | 6.5% | Điền mode (`"10+ years"`) | Biến phân loại thứ tự — không có ý nghĩa khi lấy trung bình; mode phản ánh hành vi phổ biến nhất của người vay Lending Club |
| `revol_util` | 0.08% | Điền median | Biến số liên tục có phân phối lệch; median bền vững hơn mean trước outlier |

**Tại sao không xóa dòng?**  
Với `emp_length`, xóa ~5.5% dòng (tương đương hàng chục nghìn quan sát) là lãng phí thông tin đáng kể. Điền bằng mode an toàn hơn vì biến này không phải biến dự đoán mạnh nhất.

**Tại sao không dùng mean?**  
`revol_util` có một số ngoại lai cực lớn (>100%), khiến mean bị kéo lệch — median là lựa chọn robust hơn.


````python
# ─── Phát hiện & loại bỏ Ngoại lai (IQR Method) ────────────────────────────
print('=' * 50)
print(' PHÂN TÍCH NGOẠI LAI')
print('=' * 50)

# Các biến tài chính dễ có ngoại lai
outlier_cols = ['annual_inc', 'dti', 'revol_bal', 'loan_amnt']

# ─── Boxplot trước khi xử lý ────────────────────────────────────────────────
fig, axes = plt.subplots(1, len(outlier_cols), figsize=(14, 4))
for ax, col in zip(axes, outlier_cols):
    ax.boxplot(df[col].dropna(), vert=True, patch_artist=True,
               boxprops=dict(facecolor='#AED6F1', color='#2980B9'),
               medianprops=dict(color='red', linewidth=2))
    ax.set_title(col, fontsize=10, fontweight='bold')
    ax.set_xticklabels([])
fig.suptitle('Boxplot các biến tài chính — Trước khi xử lý ngoại lai', fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()

# ─── Áp dụng IQR Capping (Winsorizing) ─────────────────────────────────────
# Dùng Capping thay vì xóa để giữ số lượng mẫu
n_before = len(df)
print(f'\nSố dòng trước: {n_before:,}')

def iqr_filter(dataframe, col):
    """Xóa các dòng nằm ngoài [Q1 - 1.5*IQR, Q3 + 1.5*IQR] cho biến col."""
    Q1 = dataframe[col].quantile(0.25)
    Q3 = dataframe[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    before = len(dataframe)
    dataframe = dataframe[(dataframe[col] >= lower) & (dataframe[col] <= upper)]
    removed = before - len(dataframe)
    print(f'   {col:15s}: lower={lower:>10.1f}, upper={upper:>10.1f} → loại {removed:,} dòng')
    return dataframe

print('\n Chi tiết loại bỏ ngoại lai:')
for col in outlier_cols:
    df = iqr_filter(df, col)

n_after = len(df)
print(f'\nSố dòng sau:  {n_after:,}')
print(f'Đã loại bỏ:   {n_before - n_after:,} dòng ({(n_before-n_after)/n_before*100:.2f}%)')

# ─── Boxplot sau khi xử lý ──────────────────────────────────────────────────
fig, axes = plt.subplots(1, len(outlier_cols), figsize=(14, 4))
for ax, col in zip(axes, outlier_cols):
    ax.boxplot(df[col].dropna(), vert=True, patch_artist=True,
               boxprops=dict(facecolor='#A9DFBF', color='#1E8449'),
               medianprops=dict(color='red', linewidth=2))
    ax.set_title(col, fontsize=10, fontweight='bold')
    ax.set_xticklabels([])
fig.suptitle('Boxplot các biến tài chính — Sau khi xử lý ngoại lai', fontweight='bold', y=1.02)
plt.tight_layout()
plt.show()
````

### – Tác động của ngoại lai đến mô hình hồi quy

**Thứ nhất ngoại lai nguy hiểm với OLS?**  
Hồi quy tuyến tính OLS tối thiểu tổng bình phương sai số (SSE). Một điểm ngoại lai duy nhất có thể kéo lệch toàn bộ mặt phẳng hồi quy vì phần dư của nó được bình phương, làm phóng đại ảnh hưởng lên nhiều lần.

**Thứ hai các biến xử lý & lý do:**

| Biến | Vấn đề | Ý nghĩa tài chính |
|------|--------|-------------------|
| `annual_inc` | Một số người khai thu nhập >$1M (cực hiếm, có thể sai số nhập liệu) | Thu nhập quá cao làm mô hình hiểu sai mối quan hệ thu nhập – lãi suất |
| `dti` | Một số khoản vay có dti > 45 (nợ gần gấp đôi thu nhập) | Cực kỳ rủi ro, không phản ánh hành vi thông thường của thị trường |
| `revol_bal` | Số dư xoay vòng cực lớn (triệu USD) | Bất thường, có thể là lỗi dữ liệu |
| `loan_amnt` | Phân phối khá đều nhưng vẫn kiểm tra đề phòng | Giữ tương đối khớp với pipeline |

**Phương pháp: Loại bỏ dòng (IQR Filtering)**  
Nhóm ưu tiên dùng phương pháp IQR vì nó là non-parametric, không bắt buộc dữ liệu phải theo phân phối chuẩn (rất hợp với dữ liệu tài chính hay bị lệch phải). Tỷ lệ loại bỏ thường < 5% — không làm mất đáng kể thông tin.


````python
# ─── Mã hóa biến phân loại (Encoding) ──────────────────────────────────────
print('=' * 50)
print(' MÃ HÓA BIẾN PHÂN LOẠI')
print('=' * 50)
print()

# ── 1. ORDINAL ENCODING (Thứ tự có ý nghĩa) ─────────────────────────────────

# term: '36 months' → 36, '60 months' → 60
df['term'] = df['term'].str.strip().str.extract(r'(\d+)').astype(int)
print(' term        → Ordinal: số tháng thực (36 / 60)')

# grade: A(1) → G(7) — thứ tự tín dụng có ý nghĩa rõ ràng
grade_map = {'A': 1, 'B': 2, 'C': 3, 'D': 4, 'E': 5, 'F': 6, 'G': 7}
df['grade'] = df['grade'].map(grade_map)
print(' grade       → Ordinal: A=1, B=2, ..., G=7')

# emp_length: thứ tự thâm niên có ý nghĩa
emp_map = {
    '< 1 year': 0, '1 year': 1, '2 years': 2, '3 years': 3,
    '4 years': 4, '5 years': 5, '6 years': 6, '7 years': 7,
    '8 years': 8, '9 years': 9, '10+ years': 10
}
df['emp_length'] = df['emp_length'].map(emp_map)
print(' emp_length  → Ordinal: "< 1 year"=0, ..., "10+ years"=10')

print()

# ── 2. ONE-HOT ENCODING (Không có thứ tự) ───────────────────────────────────
ohe_cols = ['home_ownership', 'verification_status', 'purpose']
df = pd.get_dummies(df, columns=ohe_cols, drop_first=True, dtype=int)
print(f' One-Hot Encoding: {ohe_cols}')
print(f'   → Tạo ra {df.shape[1] - 16} cột mới (drop_first=True để tránh dummy trap)')

print()
print(f' Kích thước sau Encoding: {df.shape}')
print(f'   Tất cả biến đã là kiểu số: {df.dtypes.eq(object).sum() == 0}')

# ── Thêm cột fico trung bình ────────────────────────────────────────────────
df['fico_avg'] = (df['fico_range_low'] + df['fico_range_high']) / 2
df.drop(columns=['fico_range_low', 'fico_range_high'], inplace=True)
print(' fico_avg    → Tạo điểm FICO trung bình, bỏ hai cột gốc')

# ── Xem kết quả ─────────────────────────────────────────────────────────────
print(f'\n Dataset cuối sau tiền xử lý: {df.shape}')
df.head(3)
````

###  Lý do chọn từng phương pháp Encoding

Không có một phương pháp encoding nào phù hợp cho mọi biến. Quyết định encoding phải dựa vào bản chất tài chính của từng biến:

**Ordinal Encoding** – Dùng khi biến có thứ bậc có ý nghĩa:

| Biến | Lý do dùng Ordinal |
|------|--------------------|
| `grade` (A→G) | Lending Club tự định nghĩa A = tốt nhất, G = rủi ro nhất. Thứ tự này trực tiếp ánh xạ lên lãi suất (tương quan ~0.95) — mã hóa số nguyên 1–7 là tự nhiên và hiệu quả |
| `emp_length` | Thâm niên càng cao → ổn định thu nhập càng cao → thứ tự có ý nghĩa kinh tế rõ ràng |
| `term` | Chỉ có 2 giá trị (36/60 tháng) — rút số nguyên thực là tự nhiên nhất |

**One-Hot Encoding** – Dùng khi biến danh nghĩa, không có thứ tự:

| Biến | Lý do dùng OHE |
|------|----------------|
| `home_ownership` | RENT, OWN, MORTGAGE không có thứ bậc hơn kém rõ ràng |
| `verification_status` | 3 trạng thái độc lập — mã hóa số sẽ tạo ra thứ tự giả tạo |
| `purpose` | 12 mục đích vay khác nhau, hoàn toàn danh nghĩa |

> `drop_first=True` được dùng để loại bỏ 1 dummy biến khỏi mỗi nhóm — tránh hiện tượng Dummy Variable Trap (đa cộng tuyến hoàn hảo giữa các biến dummy).

````python
# --- 2.4 PCA 2D – tô màu theo dải lãi suất ---
# Loại bỏ các dòng còn thiếu giá trị (NaN) trước khi PCA
df = df.dropna()

feature_cols = [c for c in df.columns if c != TARGET]
X_all = df[feature_cols].values
y_all = df[TARGET].values

scaler_pca = StandardScaler()
X_scaled = scaler_pca.fit_transform(X_all)

pca2 = PCA(n_components=2, random_state=RANDOM_STATE)
X_pca = pca2.fit_transform(X_scaled)

print(f'Phương sai giải thích bởi PC1+PC2: {pca2.explained_variance_ratio_.sum()*100:.2f}%')

# Chia dải lãi suất
rate_bins = pd.cut(y_all, bins=[0, 10, 15, 20, 100],
                   labels=['<10%', '10-15%', '15-20%', '>20%'])

colors_map = {'<10%': '#2196F3', '10-15%': '#4CAF50', '15-20%': '#FF9800', '>20%': '#F44336'}

idx_s = np.random.choice(len(X_pca), 8000, replace=False)

plt.figure(figsize=(10, 7))
for label, color in colors_map.items():
    mask = rate_bins[idx_s] == label
    plt.scatter(X_pca[idx_s][mask, 0], X_pca[idx_s][mask, 1],
                alpha=0.35, s=8, label=label, color=color)
plt.xlabel(f'PC1 ({pca2.explained_variance_ratio_[0]*100:.1f}%)')
plt.ylabel(f'PC2 ({pca2.explained_variance_ratio_[1]*100:.1f}%)')
plt.title('PCA 2D – Phân cụm theo dải lãi suất')
plt.legend(title='int_rate')
plt.tight_layout()
plt.show()
````

###  Nhận xét EDA: PCA 2D – Phân cụm theo dải lãi suất

**Phương sai giải thích: PC1 + PC2 = 17.25%** (thấp, do không gian đặc trưng rất đa dạng sau One-Hot Encoding). Tuy nhiên biểu đồ PCA 2D vẫn cho thấy cấu trúc phân cụm rõ ràng:

- **Dải lãi suất cao (>20%, đỏ)** tập trung ở vùng PC1 dương lớn → PC1 phản ánh thành phần "rủi ro" – `grade` cao, `fico_avg` thấp.
- **Dải lãi suất thấp (<10%, xanh dương)** nằm ở phía PC1 âm → hồ sơ tín dụng tốt.
- Dải 10–15% và 15–20% **chồng lấp nhiều** → ranh giới giữa các nhóm rủi ro trung bình không rõ ràng, gợi ý quan hệ phi tuyến.

**Hướng xử lý tiếp theo của nhóm:** Sự chồng lấp giữa các dải lãi suất trung bình cho thấy mô hình tuyến tính có thể gặp khó khăn trong việc phân biệt chính xác; cần thử nghiệm mô hình phi tuyến (Random Forest, XGBoost) để cải thiện khả năng phân tách.

## 3. XÂY DỰNG MÔ HÌNH HỒI QUY TUYẾN TÍNH ĐA BIẾN

Từ EDA, nhóm thấy dữ liệu được phân tầng rõ theo mức rủi ro tín dụng. Do đó, nhóm chọn mô hình hồi quy tuyến tính đa biến (Multiple Linear Regression) làm baseline cơ bản để so sánh, cũng như để dễ dàng giải thích mối quan hệ giữa các biến:

| Thành phần mô hình | Mục đích phân tích | Rủi ro nếu bỏ qua |
|:---|:---|:---|
| **Mô hình OLS (Ordinary Least Squares)** | Ước lượng hệ số hồi quy, đánh giá ý nghĩa thống kê từng biến | Không có baseline interpretable, khó giải thích ảnh hưởng từng yếu tố đến lãi suất |
| **Kiểm định giả định** (Shapiro-Wilk, VIF) | Đảm bảo mô hình hợp lệ, dò tìm đa cộng tuyến | Dễ dẫn đến kết luận sai lệch nếu vi phạm giả định OLS (chuẩn, độc lập, phương sai sai số đồng nhất) |
| **Regularization** (Ridge, Lasso) | Giúp xử lý đa cộng tuyến và tự động chọn lọc đặc trưng | Hệ số dễ bị mất ổn định, có nguy cơ overfitting |
| **K-Fold Cross Validation** | Đánh giá mức độ tổng quát, kiểm tra xem có over/underfitting không | Nếu chỉ đánh giá trên một tập test thì dễ sinh ra bias, không đo lường được variance của mô hình |

Nhóm tiến hành xây dựng mô hình theo trình tự: **OLS Baseline → Kiểm định giả định → Regularization → Cross-Validation → Đánh giá tổng thể**.

###  Tại sao cần kiểm tra giả định mô hình OLS?

Mô hình OLS chỉ cho kết quả **BLUE** (Best Linear Unbiased Estimator) khi thỏa mãn các giả định Gauss-Markov. Nếu vi phạm, hệ số hồi quy có thể vẫn đúng nhưng sai số chuẩn, kiểm định t, khoảng tin cậy sẽ không đáng tin.

**Theo yêu cầu đề bài**, cần kiểm tra:
1. **Phân phối chuẩn phần dư** → Histogram, Q-Q plot, Shapiro-Wilk

**Mở rộng thêm** (ngoài yêu cầu đề, để phân tích toàn diện hơn):

2. **Phương sai đồng nhất** → Residuals vs Fitted plot + Breusch-Pagan test — nếu phương sai phần dư thay đổi theo giá trị dự đoán, sai số chuẩn OLS sẽ bị ước lượng sai.

3. **Không tự tương quan phần dư** → Durbin-Watson (đã có sẵn trong OLS summary) — nếu phần dư có tự tương quan, OLS đánh giá thấp phương sai → kiểm định t bị thổi phồng.

````python
# --- 3b. Kiểm tra giả định mô hình ---
from statsmodels.stats.diagnostic import het_breuschpagan

residuals = y_train - ols_model.predict(X_train_sm)

fig, axes = plt.subplots(1, 3, figsize=(16, 5))

# Histogram phần dư
axes[0].hist(residuals, bins=80, color='steelblue', edgecolor='white', alpha=0.8)
axes[0].set_title('Phân phối phần dư (Histogram)')
axes[0].set_xlabel('Residuals')

# Q-Q plot
stats.probplot(residuals, dist='norm', plot=axes[1])
axes[1].set_title('Q-Q Plot phần dư')

# Residuals vs Fitted
fitted = ols_model.fittedvalues
axes[2].scatter(fitted, residuals, alpha=0.15, s=5, color='darkorange')
axes[2].axhline(0, color='red', linestyle='--')
axes[2].set_title('Residuals vs Fitted')
axes[2].set_xlabel('Fitted values')
axes[2].set_ylabel('Residuals')

plt.tight_layout()
plt.show()

# Kiểm định Shapiro-Wilk trên mẫu nhỏ (SW yêu cầu n ≤ 5000)
sw_sample = residuals.sample(min(5000, len(residuals)), random_state=RANDOM_STATE)
stat_sw, p_sw = stats.shapiro(sw_sample)
print(f'Shapiro-Wilk: W={stat_sw:.6f}, p-value={p_sw:.6f}')
print('→ Phần dư', 'CÓ phân phối chuẩn' if p_sw > 0.05 else 'KHÔNG có phân phối chuẩn (p < 0.05)')

# Kiểm định Breusch-Pagan (Homoscedasticity)
bp_stat, bp_p, bp_f, bp_fp = het_breuschpagan(residuals, X_train_sm)
print(f'\nBreusch-Pagan: LM={bp_stat:.4f}, p-value={bp_p:.6f}')
print('→ Phương sai phần dư', 'ĐỒNG NHẤT (homoscedastic)' if bp_p > 0.05 else 'KHÔNG đồng nhất (heteroscedastic, p < 0.05)')
````

### Đánh giá của nhóm: Kiểm định giả định mô hình OLS

#### 1. Phân phối chuẩn phần dư (Normality)

**Shapiro-Wilk:** W ≈ 0.978, p-value < 0.0001 → **Bác bỏ H₀** (phần dư không phân phối chuẩn).

- Nhóm cho rằng kết quả này là bình thường với cỡ mẫu lớn (n ≈ 160,000): Shapiro-Wilk có power rất cao nên phát hiện cả độ lệch nhỏ khỏi chuẩn.
- **Histogram**: Phần dư có dạng chuông đối xứng, gần giống phân phối chuẩn về mặt trực quan.
- **Q-Q Plot**: Phần lớn điểm bám sát đường chéo, chỉ lệch nhẹ ở hai đuôi → đuôi dày hơn phân phối chuẩn (leptokurtic), phù hợp với Kurtosis = 4.869 ở OLS summary.
- **Thực tế**: Với n lớn, CLT đảm bảo ước lượng OLS vẫn nhất quán – kiểm định t và F vẫn xấp xỉ đúng.

#### 2. Phương sai không đổi (Homoscedasticity)

**Breusch-Pagan test:** LM ≈ 17733.46, p-value < 0.05 → **Bác bỏ H₀** → phần dư có **heteroscedasticity** (phương sai không đồng nhất).

- **Residuals vs Fitted plot** xác nhận: độ phân tán phần dư thay đổi theo giá trị fitted – phần dư rộng hơn ở vùng lãi suất thấp và cao, hẹp hơn ở vùng trung bình.
- Heteroscedasticity **không** làm hệ số OLS bị chệch (biased), nhưng sai số chuẩn (standard errors) có thể bị sai → ảnh hưởng đến kiểm định t và khoảng tin cậy.
- **Giải pháp**: Có thể dùng Robust Standard Errors (HC3) trong statsmodels, hoặc chuyển sang mô hình phi tuyến (RF, XGBoost) không yêu cầu giả định homoscedasticity.

#### 3. Không có tự tương quan (Independence)

**Durbin-Watson = 1.997** (từ OLS summary) → Xấp xỉ 2 → **Không có tự tương quan phần dư**

#### Tổng hợp lại các đánh giá

| Giả định | Kiểm định | Kết quả |
|-----------|-----------|---------|
| Phân phối chuẩn phần dư | Shapiro-Wilk (W=0.978) |  Vi phạm (nhưng chấp nhận được với n lớn – CLT) |
| Phương sai đồng nhất | Breusch-Pagan (LM=17733) |  Vi phạm → cần robust SE hoặc mô hình phi tuyến |
| Không tự tương quan | Durbin-Watson (DW=1.997) |  Thỏa mãn |
| Không đa cộng tuyến | VIF |  Kiểm tra ở bước tiếp theo |

Mô hình OLS cho R² cao (~0.917) nhưng vi phạm 2 giả định. Điều này khuyến nghị sử dụng thêm các mô hình phi tham số (RF, XGBoost) và regularization (Ridge, Lasso) để tăng độ tin cậy.

````python
# Đa cộng tuyến – Hệ số VIF
X_vif = X_train_sm.copy()

vif_data = pd.DataFrame()
vif_data['feature'] = X_vif.columns
vif_data['VIF'] = [variance_inflation_factor(X_vif.values, i) for i in range(X_vif.shape[1])]
vif_data = vif_data.sort_values('VIF', ascending=False).reset_index(drop=True)

print('Hệ số VIF:')
print(vif_data.to_string())

plt.figure(figsize=(10, 6))
top_vif = vif_data[vif_data['feature'] != 'const'].head(20)
plt.barh(top_vif['feature'][::-1], top_vif['VIF'][::-1], color='steelblue', edgecolor='white')
plt.axvline(10, color='red', linestyle='--', label='VIF=10 (ngưỡng cao)')
plt.axvline(5, color='orange', linestyle='--', label='VIF=5 (ngưỡng trung bình)')
plt.title('Hệ số VIF – Kiểm tra đa cộng tuyến')
plt.xlabel('VIF')
plt.legend()
plt.tight_layout()
plt.show()
````

### Đánh giá của nhóm: Kiểm tra Đa cộng tuyến (VIF)

**Kết quả VIF (sau khi loại const):**

**Nhóm nhận thấy:**
- **`fico_avg`** có VIF thấp — biến FICO trung bình duy nhất trong mô hình, không gây đa cộng tuyến.

**Hướng xử lý:** Cân nhắc loại `installment` hoặc `loan_amnt` để cải thiện tính ổn định của mô hình. Ridge/Lasso ở bước tiếp theo sẽ giúp kiểm soát VIF cao tự nhiên.

````python
# --- 3c. Tạo biến mới (Interaction terms + log transform) ---
def add_engineered_features(df_in):
    df_out = df_in.copy()
    df_out['dti_x_annual_inc'] = df_out['dti'] * df_out['annual_inc']          # Tương tác: DTI × Thu nhập
    df_out['log_loan_amnt']    = np.log1p(df_out['loan_amnt'])                 # Log số tiền vay
    df_out['log_annual_inc']   = np.log1p(df_out['annual_inc'])                # Log thu nhập
    df_out['grade_x_dti']      = df_out['grade'] * df_out['dti']               # Tương tác: Grade × DTI
    return df_out

X_train_eng = add_engineered_features(X_train)
X_test_eng  = add_engineered_features(X_test)

print(f'Số đặc trưng sau feature engineering: {X_train_eng.shape[1]}')

# Huấn luyện lại OLS với biến mới
X_train_eng_sm = sm.add_constant(X_train_eng)
X_test_eng_sm  = sm.add_constant(X_test_eng)

ols_eng = sm.OLS(y_train, X_train_eng_sm).fit()
y_pred_eng = ols_eng.predict(X_test_eng_sm)

r2_eng  = r2_score(y_test, y_pred_eng)
mae_eng = mean_absolute_error(y_test, y_pred_eng)
rmse_eng = np.sqrt(mean_squared_error(y_test, y_pred_eng))
print(f'OLS+Engineering – R²={r2_eng:.4f}  MAE={mae_eng:.4f}  RMSE={rmse_eng:.4f}')
````

````python
# --- 3d. Ridge và Lasso với Cross-Validation tuning ---
from sklearn.linear_model import RidgeCV, LassoCV

scaler = StandardScaler()
X_train_sc = scaler.fit_transform(X_train_eng)
X_test_sc  = scaler.transform(X_test_eng)

alphas = np.logspace(-3, 3, 50)

# Ridge
ridge_cv = RidgeCV(alphas=alphas, cv=5)
ridge_cv.fit(X_train_sc, y_train)
y_pred_ridge = ridge_cv.predict(X_test_sc)
r2_ridge  = r2_score(y_test, y_pred_ridge)
mae_ridge = mean_absolute_error(y_test, y_pred_ridge)
rmse_ridge = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
print(f'Ridge (alpha={ridge_cv.alpha_:.4f}) – R²={r2_ridge:.4f}  MAE={mae_ridge:.4f}  RMSE={rmse_ridge:.4f}')

# Lasso
lasso_cv = LassoCV(alphas=alphas, cv=5, max_iter=10000, random_state=RANDOM_STATE)
lasso_cv.fit(X_train_sc, y_train)
y_pred_lasso = lasso_cv.predict(X_test_sc)
r2_lasso  = r2_score(y_test, y_pred_lasso)
mae_lasso = mean_absolute_error(y_test, y_pred_lasso)
rmse_lasso = np.sqrt(mean_squared_error(y_test, y_pred_lasso))
print(f'Lasso (alpha={lasso_cv.alpha_:.6f}) – R²={r2_lasso:.4f}  MAE={mae_lasso:.4f}  RMSE={rmse_lasso:.4f}')

# Hệ số bị triệt tiêu bởi Lasso
n_zero = np.sum(lasso_cv.coef_ == 0)
print(f'Lasso đã triệt tiêu {n_zero}/{len(lasso_cv.coef_)} hệ số → chọn lọc đặc trưng tự động.')
````

### Đánh giá của nhóm: Ridge & Lasso Regression

| Mô hình | Alpha tối ưu | R² | MAE | RMSE |
|---------|-------------|-----|-----|------|

| Lasso | 0.0010 | 0.9160 | 1.0689 | 1.3879 |

**Nhóm nhận thấy:**
- **Alpha tối ưu nhỏ** cho cả Ridge (0.28) và Lasso (0.001) → mức độ regularization nhẹ là đủ; dữ liệu lớn (~160K training rows) giúp mô hình tuyến tính tự nhiên không bị overfitting nghiêm trọng.
- Ridge và Lasso cho **R² = 0.916** – cải thiện nhẹ so với OLS gốc (0.9145) nhờ regularization kiểm soát đa cộng tuyến (đặc biệt home_ownership và loan_amnt/installment).
- **Lasso triệt tiêu 5/40 hệ số về 0**: tự động loại bỏ 5 đặc trưng ít quan trọng, giúp mô hình đơn giản hơn (sparse solution). Đây là ưu điểm của Lasso so với Ridge trong bối cảnh tài chính cần giải thích.
- Ridge phân phối lại hệ số đồng đều hơn (L2 penalty), phù hợp khi cần giữ lại tất cả đặc trưng nhưng kiểm soát đa cộng tuyến.

**Tóm lại:** Cả hai phương pháp đều xử lý hiệu quả vấn đề đa cộng tuyến (VIF cao ở home_ownership, loan_amnt/installment). Lasso có thêm giá trị thực tiễn trong feature selection.

````python
# --- 3e. K-Fold Cross Validation (K=5) ---
from sklearn.linear_model import Ridge as RidgeSK

kf = KFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

def kfold_eval(model, X, y, kf):
    r2s, maes, rmses = [], [], []
    for train_idx, val_idx in kf.split(X):
        X_tr, X_val = X[train_idx], X[val_idx]
        y_tr, y_val = y.iloc[train_idx], y.iloc[val_idx]
        model.fit(X_tr, y_tr)
        preds = model.predict(X_val)
        r2s.append(r2_score(y_val, preds))
        maes.append(mean_absolute_error(y_val, preds))
        rmses.append(np.sqrt(mean_squared_error(y_val, preds)))
    return np.array(r2s), np.array(maes), np.array(rmses)

X_kfold = scaler.transform(add_engineered_features(X))

models_kfold = {
    'OLS (LinearRegression)': LinearRegression(),
    'Ridge': RidgeSK(alpha=ridge_cv.alpha_),
    'Lasso': lasso_cv,
}

results_kfold = {}
for name, m in models_kfold.items():
    r2s, maes, rmses = kfold_eval(m, X_kfold, y, kf)
    results_kfold[name] = {'R²': r2s.mean(), 'MAE': maes.mean(), 'RMSE': rmses.mean(),
                           'R²_std': r2s.std()}
    print(f'{name:30s} R²={r2s.mean():.4f}±{r2s.std():.4f}  MAE={maes.mean():.4f}  RMSE={rmses.mean():.4f}')

df_kfold = pd.DataFrame(results_kfold).T
print('\nBảng tổng hợp K-Fold (K=5):')
print(df_kfold.to_string())
````

| Ridge | 0.9183 | ±0.0014 | 1.0611 | 1.3722 |

**Nhóm nhận thấy:**
- **Std R² rất nhỏ (±0.0014)** trên 5 fold: các mô hình tuyến tính có tính ổn định (stability) cao – không bị overfitting và phân phối đều trên các fold.
- OLS, Ridge và Lasso cho kết quả gần như giống nhau trên K-Fold → với bộ dữ liệu lớn, regularization không mang lại cải thiện đáng kể (alpha tối ưu nhỏ).
- K-Fold xác nhận rằng R² ≈ 0.918 là ước lượng đáng tin cậy về hiệu năng tổng quát của mô hình tuyến tính – không phải kết quả may mắn từ một train/test split cụ thể.

**Tóm lại:** Các mô hình tuyến tính ổn định và tổng quát hóa tốt, nhưng giới hạn tự nhiên của chúng là ~0.918 R² – cần mô hình phi tuyến để vượt qua ngưỡng này.

## 4. SO SÁNH VỚI MÔ HÌNH HỌC MÁY PHI TUYẾN

Mặc dù OLS cho R2 khá tốt (~0.914), nhóm nhận thấy nó vẫn còn hạn chế khi xử lý các mối quan hệ phi tuyến phức tạp. Vì vậy, nhóm quyết định dùng thêm các mô hình học máy phi tuyến như Random Forest và XGBoost để dò tìm các tương tác ẩn, từ đó mong muốn cải thiện độ chính xác:

| Thành phần phân tích | Mục đích phân tích | Rủi ro nếu bỏ qua |
|:---|:---|:---|
| **Random Forest** (Bagging) | Xử lý phi tuyến, giảm overfitting qua tổng hợp nhiều cây | Bỏ lỡ mô hình robust với outliers, không tận dụng được tính đa dạng của ensemble |
| **XGBoost** (Boosting) | Học tuần tự sửa lỗi, tối ưu hóa dần dần, hiệu năng cao nhất | Không đạt được độ chính xác tối đa, bỏ qua phương pháp state-of-the-art trong credit scoring |
| **Hyperparameter Tuning** (RandomizedSearchCV) | Tìm cấu hình tối ưu cho từng mô hình, tránh overfitting/underfitting | Mô hình hoạt động ở hiệu năng sub-optimal, có thể overfit nghiêm trọng nếu cây quá sâu |
| **Feature Importance** | Xác định yếu tố ảnh hưởng mạnh nhất, so sánh nhất quán giữa mô hình | Không biết lý do tại sao mô hình đưa ra quyết định, giảm khả năng interpret và trust |
| **So sánh tổng thể** (R², MAE, RMSE) | Chọn mô hình tốt nhất cho triển khai thực tế | Không có cơ sở quyết định mô hình nào phù hợp cho ngân hàng |

Trình tự so sánh của nhóm: **Random Forest (tuning) → XGBoost (tuning) → Feature Importance → Bảng so sánh toàn diện**.

````python
# --- 4.1 Random Forest với RandomizedSearchCV ---
from scipy.stats import randint, uniform

rf_param_dist = {
    'n_estimators': [50, 100, 200],
    'max_depth': [5, 10, 15, 20],
    'min_samples_split': [2, 5, 10],
    'max_features': [0.3, 0.5, 0.7, 'sqrt'],
}

rf_search = RandomizedSearchCV(
    RandomForestRegressor(random_state=RANDOM_STATE),
    param_distributions=rf_param_dist,
    n_iter=10, cv=3, scoring='r2',
    random_state=RANDOM_STATE, n_jobs=-1, verbose=1
)
rf_search.fit(X_train_eng, y_train)
best_rf = rf_search.best_estimator_

print(f'Best RF params: {rf_search.best_params_}')

y_pred_rf = best_rf.predict(X_test_eng)
r2_rf   = r2_score(y_test, y_pred_rf)
mae_rf  = mean_absolute_error(y_test, y_pred_rf)
rmse_rf = np.sqrt(mean_squared_error(y_test, y_pred_rf))
print(f'Random Forest – R²={r2_rf:.4f}  MAE={mae_rf:.4f}  RMSE={rmse_rf:.4f}')
````

**Tóm lại:** Random Forest cho thấy bước cải thiện so với OLS, xác nhận sự tồn tại của quan hệ phi tuyến trong dữ liệu tín dụng.

````python
# --- 4.2 XGBoost với RandomizedSearchCV ---
xgb_param_dist = {
    'n_estimators': [100, 200, 300],
    'max_depth': [4, 6, 8, 10],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'subsample': [0.7, 0.8, 0.9, 1.0],
    'colsample_bytree': [0.7, 0.8, 0.9, 1.0],
}

xgb_search = RandomizedSearchCV(
    XGBRegressor(random_state=RANDOM_STATE, tree_method='hist'),
    param_distributions=xgb_param_dist,
    n_iter=10, cv=3, scoring='r2',
    random_state=RANDOM_STATE, n_jobs=-1, verbose=1
)
xgb_search.fit(X_train_eng, y_train)
best_xgb = xgb_search.best_estimator_

print(f'Best XGBoost params: {xgb_search.best_params_}')

y_pred_xgb = best_xgb.predict(X_test_eng)
r2_xgb   = r2_score(y_test, y_pred_xgb)
mae_xgb  = mean_absolute_error(y_test, y_pred_xgb)
rmse_xgb = np.sqrt(mean_squared_error(y_test, y_pred_xgb))
print(f'XGBoost – R²={r2_xgb:.4f}  MAE={mae_xgb:.4f}  RMSE={rmse_xgb:.4f}')
````

## 2. Xây dựng PCA từ đầu (From Scratch)

### Mô tả thuật toán PCA tự xây dựng (From Scratch)

PCA được xây dựng hoàn toàn từ đầu bằng NumPy, **không dùng** `sklearn.decomposition.PCA`, theo 4 bước:

| Bước | Thao tác | Công thức / Hàm |
|------|----------|-----------------|
| 1 | **Chuẩn hóa (Zero-Mean)** | X_centered = X − mean(X, axis=0) |
| 2️ | **Ma trận hiệp phương sai** | Σ = X_centered^T · X_centered / (T−1) |
| 3️ | **Phân tích trị riêng** | Σ · v = λ · v  →  dùng `np.linalg.eigh` |
| 4️ | **Chiếu lên PC** | PC_scores = X_centered · V (eigenvectors) |

**Lý thuyết nền:**
- Ma trận hiệp phương sai Σ là ma trận đối xứng xác định dương (PSD) → tất cả trị riêng λ ≥ 0.
- Vector riêng (eigenvector) tương ứng với trị riêng lớn nhất chỉ hướng có phương sai lớn nhất → đây là PC1.
- Các vector riêng trực giao nhau → các PC độc lập tuyến tính với nhau.
- `np.linalg.eigh` (dành cho ma trận Hermitian/symmetric) cho kết quả chính xác hơn `eig` tổng quát và trả về trị riêng theo thứ tự tăng dần → cần đảo ngược index để PC1 có phương sai lớn nhất.

````python
# --- Bước 1: Chuẩn hóa (zero-mean) ---
R = returns.values                          # (T, N)
R_mean = R.mean(axis=0)                     # (N,)
R_centered = R - R_mean                     # Trừ mean từng cổ phiếu

print(f'Ma trận return chuẩn hóa: {R_centered.shape}')
print(f'Mean sau chuẩn hóa (kiểm tra ≈ 0): {np.abs(R_centered.mean(axis=0)).max():.2e}')
````

### Đánh giá của nhóm: Bước 1 – Chuẩn hóa dữ liệu (Zero-Mean Centering)

**Lý do chuẩn hóa trong PCA:**
- PCA tìm các hướng có phương sai lớn nhất trong không gian đặc trưng. Phương sai được tính quanh giá trị trung bình.
- Nếu không trừ trung bình, PC1 sẽ bị "kéo" về hướng của drift (xu hướng tăng/giảm trung bình) thay vì nắm bắt cấu trúc đồng biến động thực sự.

**Kỹ thuật:** Trong bài này dùng **Zero-Mean Centering** (chỉ trừ trung bình, không chia std) vì các cổ phiếu đều là tỷ suất sinh lợi (return) – cùng đơn vị (%) – nên không cần chuẩn hóa về cùng scale. Nếu dùng giá tuyệt đối, cần chuẩn hóa Z-score (trừ mean, chia std) để tránh cổ phiếu giá cao (VIC, VHM) chi phối PCA.

**Kiểm tra:** Mean sau centering ≈ 9.14e-18 ≈ 0  → bước chuẩn hóa thành công.

````python
# --- Bước 2: Ma trận hiệp phương sai ---
T = R_centered.shape[0]
cov_matrix = (R_centered.T @ R_centered) / (T - 1)   # (N, N)

print(f'Ma trận hiệp phương sai: {cov_matrix.shape}')

# Trực quan hóa
plt.figure(figsize=(12, 10))
sns.heatmap(pd.DataFrame(cov_matrix, index=returns.columns, columns=returns.columns),
            cmap='RdBu_r', center=0, annot=False, fmt='.4f')
plt.title('Ma trận Hiệp phương sai – Return hàng ngày VN30')
plt.tight_layout()
plt.show()
````

### Đánh giá của nhóm: Bước 2 – Ma trận Hiệp phương sai (Covariance Matrix)

**Cách tính:** Σ = (X_centered^T × X_centered) / (T − 1), trong đó T là số phiên giao dịch. Kết quả là ma trận **đối xứng 30×30**, mỗi phần tử Σᵢⱼ = Cov(stock_i, stock_j).

**Nhận xét Heatmap:**
- Phần lớn các ô trên heatmap có màu **ấm (dương)** → hầu hết cặp cổ phiếu VN30 có hiệp phương sai dương → **cùng chiều biến động**.
- Đường chéo chính (variance) có giá trị cao nhất, các cổ phiếu có thanh khoản thấp hoặc biến động mạnh có variance lớn hơn.
- Một số cặp có hiệp phương sai thấp hoặc gần 0 → ít đồng biến động (potential hedging pairs).

**Điều này có nghĩa là:** Ma trận hiệp phương sai dương (gần như) xác định (positive semi-definite) là điều kiện cần để PCA hoạt động đúng. Mức hiệp phương sai cao phổ biến là dấu hiệu của **systemic risk cao** trong thị trường VN30 – các cổ phiếu ít có tác dụng đa dạng hóa rủi ro trong điều kiện thị trường xấu.

````python
# --- Bước 3: Tính trị riêng và vector riêng ---
eigenvalues, eigenvectors = np.linalg.eigh(cov_matrix)  # eigh cho ma trận đối xứng

# Sắp xếp giảm dần
order = np.argsort(eigenvalues)[::-1]
eigenvalues  = eigenvalues[order]
eigenvectors = eigenvectors[:, order]   # Mỗi cột là 1 eigenvector

# Tỷ lệ phương sai giải thích
explained_ratio = eigenvalues / eigenvalues.sum()
cum_explained   = np.cumsum(explained_ratio)

print('Top 10 eigenvalues:')
for i in range(10):
    print(f'  PC{i+1}: λ={eigenvalues[i]:.6f}  Var={explained_ratio[i]*100:.2f}%  Cum={cum_explained[i]*100:.2f}%')
````

### Đánh giá của nhóm: Bước 3 – Trị riêng & Vector riêng (Eigendecomposition)

**Kết quả phân tích:**
- **PC1: λ ≈ 13.5, giải thích 45.12%** tổng phương sai → một yếu tố hệ thống đơn lẻ chi phối phần lớn biến động của rổ VN30.
- PC2 (~7%), PC3 (~5%), ...: mỗi thành phần giải thích < 7%, đại diện cho các yếu tố nhỏ hơn (ngành, nhóm).
- Phương sai tích lũy PC1–PC3 ≈ 57%; PC1–PC5 ≈ 70%.

**Lý do dùng `numpy.linalg.eigh`:** Hàm `eigh` được tối ưu hóa cho ma trận đối xứng thực (symmetric real matrix)  đúng với ma trận hiệp phương sai cho kết quả chính xác và nhanh hơn `eig` tổng quát. Trị riêng được sắp xếp tăng dần, cần đảo ngược để lấy PC1 là PC có phương sai lớn nhất.

**Điều này có nghĩa là:** Kết quả eigendecomposition cho thấy thị trường VN30 có cấu trúc một nhân tố thống trị (single-factor dominated), tương đối khớp với mô hình định giá tài sản CAPM trong thị trường mới nổi.

````python
# Scree Plot
n_show = min(15, len(eigenvalues))
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].bar(range(1, n_show + 1), explained_ratio[:n_show] * 100, color='steelblue', edgecolor='white')
axes[0].set_title('Scree Plot – Phương sai giải thích theo từng PC')
axes[0].set_xlabel('Thành phần chính (PC)')
axes[0].set_ylabel('% Phương sai giải thích')

axes[1].plot(range(1, n_show + 1), cum_explained[:n_show] * 100, 'o-', color='darkorange')
axes[1].axhline(80, color='red', linestyle='--', label='80%')
axes[1].set_title('Phương sai tích lũy')
axes[1].set_xlabel('Số thành phần chính')
axes[1].set_ylabel('% Phương sai tích lũy')
axes[1].legend()

plt.tight_layout()
plt.show()
````

````python
# --- Biplot: PCA Scores và Loadings (VN30) ---
# Tính PCA scores cho PC1 và PC2
pca_scores = R_centered @ eigenvectors[:, :2]

# Tính loadings (tích eigenvector với sqrt(eigenvalue) để có tỷ lệ chính xác)
loadings = eigenvectors[:, :2] * np.sqrt(eigenvalues[:2])

plt.figure(figsize=(12, 10))

# Vẽ các điểm dữ liệu (PCA scores)
plt.scatter(pca_scores[:, 0], pca_scores[:, 1], alpha=0.4, s=20, color='steelblue', edgecolors='white', linewidth=0.5)

# Vẽ các mũi tên loadings cho từng cổ phiếu
for i, ticker in enumerate(returns.columns):
    # Vẽ mũi tên
    plt.arrow(0, 0, loadings[i, 0], loadings[i, 1], 
              color='darkorange', alpha=0.7, head_width=0.03, linewidth=1.5)
    # Vẽ ticker
    plt.text(loadings[i, 0] * 1.15, loadings[i, 1] * 1.15, 
             ticker, color='darkred', fontsize=9, ha='center', va='center', fontweight='bold')

# Định dạng biểu đồ
plt.axhline(y=0, color='black', linestyle='-', linewidth=1)
plt.axvline(x=0, color='black', linestyle='-', linewidth=1)
plt.xlabel(f'PC1 ({explained_ratio[0]*100:.1f}% phương sai)', fontsize=12)
plt.ylabel(f'PC2 ({explained_ratio[1]*100:.1f}% phương sai)', fontsize=12)
plt.title('Biplot: PCA Scores và Loadings – Rổ VN30', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.2)
plt.tight_layout()
plt.show()
````

## 3. Phân tích và trực quan hóa PC1

````python
# --- Xây dựng PC1_index ---
pc1_vector = eigenvectors[:, 0]          # Vector riêng của PC1 (N,)
PC1_scores = R_centered @ pc1_vector     # (T,) – chiếu lên PC1

PC1_series = pd.Series(PC1_scores, index=returns.index, name='PC1')

# Trọng số của các cổ phiếu trong PC1
pc1_weights = pd.Series(pc1_vector, index=returns.columns, name='PC1_weight')
pc1_weights_sorted = pc1_weights.sort_values(ascending=False)

plt.figure(figsize=(12, 5))
colors_w = ['steelblue' if w > 0 else 'tomato' for w in pc1_weights_sorted]
plt.bar(pc1_weights_sorted.index, pc1_weights_sorted.values, color=colors_w, edgecolor='white')
plt.title('Trọng số các cổ phiếu VN30 trong PC1 (Eigenvector)')
plt.xlabel('Mã cổ phiếu')
plt.ylabel('Trọng số PC1')
plt.xticks(rotation=45)
plt.axhline(0, color='black', linewidth=0.8)
plt.tight_layout()
plt.show()

print('Trọng số PC1 (sorted):')
print(pc1_weights_sorted.to_string())
````

### Đánh giá của nhóm: Trọng số các cổ phiếu VN30 trong PC1 (Eigenvector)

**Điểm nổi bật nhóm thấy:**
- **Tất cả trọng số đều dương** → PC1 đại diện cho *yếu tố thị trường chung (Market Factor): khi thị trường tăng, tất cả cổ phiếu đều tăng cùng chiều với PC1.
- Top trọng số cao nhất: **GVR (0.268), SSI (0.235), POW (0.219), MSN (0.214)** → những cổ phiếu này co-movement mạnh nhất với xu hướng chung của VN30.
- Trọng số thấp nhất: **SSB (0.076)** → ít bị ảnh hưởng nhất bởi yếu tố hệ thống, có thể do đặc thù kinh doanh hoặc thanh khoản thấp.

**Ý nghĩa tài chính (theo lý thuyết CAPM):**
- Cổ phiếu có trọng số PC1 **cao** → Beta cao → rủi ro hệ thống lớn, kỳ vọng sinh lợi cao hơn trong thị trường tăng giá.
- Cổ phiếu có trọng số PC1 **thấp** → Beta thấp → phòng thủ tốt hơn trong thị trường giảm giá.
- Nhà đầu tư có thể dùng trọng số PC1 để xây dựng danh mục index-tracking với ít cổ phiếu nhất mà vẫn nắm bắt được xu hướng thị trường.

````python
# --- Cumulative return tích lũy PC1_index vs VN30 ---

# PC1 index – tích lũy theo kiểu cộng dồn (normalized)
PC1_cumsum = PC1_series.cumsum()
PC1_norm   = (PC1_cumsum - PC1_cumsum.iloc[0]) / PC1_cumsum.std()

# VN30-Index – chuẩn hóa tương tự
vn30_cumret = (1 + vn30_idx_ret).cumprod()
vn30_norm   = (vn30_cumret - vn30_cumret.iloc[0]) / vn30_cumret.std()

fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# Plot 1: Chuỗi thời gian chuẩn hóa
axes[0].plot(PC1_norm.index, PC1_norm.values, label='PC1_index (chuẩn hóa)',
             color='steelblue', linewidth=1.5)
axes[0].plot(vn30_norm.index, vn30_norm.values, label='VN30-Index (ETF proxy, chuẩn hóa)',
             color='darkorange', linewidth=1.5, linestyle='--')
axes[0].set_title('So sánh PC1_index và VN30-Index (chuỗi thời gian chuẩn hóa)')
axes[0].set_ylabel('Giá trị chuẩn hóa')
axes[0].legend()
axes[0].set_xlabel('Ngày')

# Plot 2: Scatter correlation
common = PC1_series.index.intersection(vn30_idx_ret.index)
pc1_aligned  = PC1_series.loc[common]
vn30_aligned = vn30_idx_ret.loc[common]

axes[1].scatter(vn30_aligned, pc1_aligned, alpha=0.4, s=15, color='steelblue')
m, b = np.polyfit(vn30_aligned, pc1_aligned, 1)
xline = np.linspace(vn30_aligned.min(), vn30_aligned.max(), 100)
axes[1].plot(xline, m*xline + b, color='red', linewidth=2, label=f'Fit: y={m:.4f}x+{b:.6f}')
axes[1].set_title('Scatter: PC1_index daily vs VN30-Index daily returns')
axes[1].set_xlabel('VN30-Index daily return')
axes[1].set_ylabel('PC1 score')
axes[1].legend()

plt.tight_layout()
plt.show()
````

### Đánh giá của nhóm: So sánh PC1_index và VN30-Index theo thời gian

**Nhận xét biểu đồ chuỗi thời gian (Plot trên):**
- PC1_index (chuẩn hóa) và VN30-Index di chuyển rất đồng bộ theo thời gian: các điểm đỉnh, đáy và điểm gãy của hai chuỗi gần như trùng nhau.
- Điều này xác nhận PC1 nắm bắt được xu hướng chung của thị trường mà không cần dùng nhãn (unsupervised).

**Nhận xét scatter plot (Plot dưới):**
- Các điểm scatter phân bố gần đường hồi quy tuyến tính → quan hệ PC1 ∼ VN30 gần tuyến tính.
- Hệ số góc (slope m > 0) → PC1 và VN30-Index cùng chiều → PC1 dương khi thị trường tăng, âm khi thị trường giảm.
- Một số outlier xuất hiện ở vùng return cực đoan (±2%), tương ứng các phiên có biến động lớn bất thường.

**Điểm đáng chú ý khi áp dụng thực tế:** PC1 có thể được dùng như một signal lọc xu hướng thị trường trong chiến lược đầu tư theo momentum.

````python
# --- Hệ số tương quan PC1 và VN30 ---
corr_val, p_corr = stats.pearsonr(pc1_aligned.values, vn30_aligned.values)
print(f'Hệ số tương quan Pearson (PC1 vs VN30-Index): r = {corr_val:.4f}  (p={p_corr:.4e})')

corr_spearman, _ = stats.spearmanr(pc1_aligned.values, vn30_aligned.values)
print(f'Hệ số tương quan Spearman: ρ = {corr_spearman:.4f}')

print(f'\nPC1 giải thích {explained_ratio[0]*100:.2f}% tổng phương sai của rổ VN30.')
````

### Đánh giá của nhóm: Hệ số tương quan PC1 – VN30-Index

- Spearman ρ = **0.803** → tương quan thứ hạng cũng cao, không bị ảnh hưởng bởi phân phối.

**Nhóm giải thích như sau:**

- Kết quả này xác nhận **PC1 ≡ Market Factor (Beta)**: PCA hoàn toàn tự nhiên hội tụ về yếu tố hệ thống chính mà không cần giám sát.
